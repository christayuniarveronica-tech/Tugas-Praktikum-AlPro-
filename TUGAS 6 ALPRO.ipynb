```python
#a. Membuat Variabel Global
skor_nobita = 0  # Menyimpan skor nobita selama simulasi berlangsung

#b. Membuat Fungsi untuk Memperbarui Skor


def update_skor(nilai):
  global skor_nobita
  skor_nobita += nilai

#c. Membuat Fungsi Aktivitas Nobita


#1. Fungsi belajar(jam_belajar)
def belajar(jam_belajar):
  if jam_belajar < 1:
    update_skor(50)
    return "Nobita membaca buku sebentar"
  elif 1 <= jam_belajar <= 3:
    update_skor(25)
    return "Nobita belajar secukupnya"
  else:  # jam_belajar > 3
    update_skor(10)
    return "Nobita belajar dengan giat hari ini"


#2. Fungsi tidur(jam_tidur)
def tidur(jam_tidur):
  if jam_tidur < 7:
    update_skor(10)
    return "Nobita kurang tidur"
  elif 7 <= jam_tidur <= 8:
    update_skor(20)
    return "Waktu tidur Nobita seimbang"
  else:  # jam_tidur > 8
    update_skor(20)
    return "Nobita tidur terlalu lama"


#3. Fungsi main bersama Doraemon (jumlah_gadget, waktu_main)
def main(jumlah_gadget, waktu_main):
  if jumlah_gadget <= 1 and waktu_main <= 2:
    update_skor(15)
    return "Nobita bermain secukupnya"
  elif jumlah_gadget <= 2 and waktu_main <= 4:
    update_skor(5)
    return "Nobita masih bisa mengatur waktu dengan baik"
  else:
    update_skor(-30)
    return "Nobita kebanyakan main"


#4. Fungsi menyontek()
def menyontek():
  update_skor(-50)
  return "Nobita ketahuan menyontek"

#Membuat Fungsi Pengendali Aktivitas


def pilih_aktivitas(pilihan):
  if pilihan == 1:
    jam = float(input("Masukkan jam belajar: "))
    return belajar(jam)
  elif pilihan == 2:
    waktu = float(input("Masukkan jam tidur: "))
    return tidur(waktu)
  elif pilihan == 3:
    jml_gadget = int(input("Masukkan jumlah gadget: "))
    wkt_main = float(input("Masukkan waktu main (jam): "))
    return main(jml_gadget, wkt_main)
  elif pilihan == 4:
    return menyontek()
  else:
    return "Pilihan tidak valid, aktivitas hanya ada 4 pilihan"

# e. Membuat Fungsi Evaluasi Akhir


def evaluasi_akhir():
  if skor_nobita >= 80:
    return "Siswa Teladan"
  elif 50 <= skor_nobita < 80:
    return "Cukup Rajin"
  else:  # < 50
    return "Pemalas"

# f. Membuat Menu Utama Program Peningkatan Disiplin Harian


def main_program():
  global skor_nobita
  skor_nobita = 0  

  while True:
    print("\n=== PROGRAM PENINGKATAN DISIPLIN NOBITA ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main bersama Doraemon")
    print("4. Menyontek saat ujian")
    print("5. Keluar & Lihat Evaluasi")

    pilihan = int(input("Pilih aktivitas (1-5): "))

    if pilihan == 5:
      print(f"\nSkor Akhir Nobita: {skor_nobita}")
      print(f"Hasil Evaluasi: {evaluasi_akhir()}")
      print("Program selesai. Terima kasih!")
      break
    hasil = pilih_aktivitas(pilihan)
    print(f"Output: {hasil}")
    print(f"Skor sementara: {skor_nobita}")
if __name__ == "__main__":
  main_program()
```
