```python
# Mencatat pengeluaran harian nobita
hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input(f"Masukkan pengeluaran hari ke-{hari}: "))
    saldo -= pengeluaran
    
    if saldo < 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
        
    print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")
    hari += 1
else:
    print("Pencatatan harian selesai.")
```

```python
# Menganalisis pengeluaran mingguan Nobita
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    minggu = i + 1
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{minggu}: Nobita tidak jajan, dilewati analisisnya.")
        continue
    else:
        print(f"Minggu ke-{minggu}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis mingguan selesai.")
```

```python
# Menyusun laporan bulanan Nobita
for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")
    total_bulanan = 0
    terlalu_besar = False
    
    for minggu in range(1, 5):
        pengeluaran = int(input(f"Masukkan pengeluaran minggu ke-{minggu}: "))
        total_bulanan += pengeluaran
        
        if pengeluaran > 20000:
            terlalu_besar = True
            break
            
    if terlalu_besar:
        print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")

print("\nSemua bulan selesai.")
```

```python
# Evaluasi Akhir Tahun
saldo = 300000
saldo_awal = saldo


for bulan in range(1, 13):
    print(f"\n--- Bulan ke-{bulan} ---")
    minggu = 1
    
    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")
        if libur == 'y':
            print(f"Minggu ini dilewati (continue).")
            minggu += 1
            continue
        elif libur == 'n':
            pengeluaran = int(input(f"Masukkan pengeluaran minggu ke-{minggu}: "))
            saldo -= pengeluaran
            
            if saldo < 0:
                break
        minggu += 1
        
    if saldo < 0:
        print("Saldo habis (< 0), simulasi berhenti.")
        break
        
    print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}")
else:
    print(f"\nNobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")
```
